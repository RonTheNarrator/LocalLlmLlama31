```python
from langchain_ollama import ChatOllama
from langchain.agents import create_agent
import random

llm = ChatOllama(
    model="llama3.1:8b",
    temperature=0,
    # other params...
)

reviews = [
    "Great quality and exactly as described.",
    "Fast shipping and excellent packaging.",
    "Very easy to use and works perfectly.",
    "Exceeded my expectations for the price.",
    "I would definitely buy this again.",

    "Poor quality and feels cheaply made.",
    "Item arrived later than expected.",
    "Did not match the product description.",
    "Stopped working after a few uses.",
    "Not worth the money."
]
random.seed(66)
random.shuffle(reviews)
reviews = [[r,None] for r in reviews]
counter = -1

def read_next_review() -> str:
    """
    Returns next review from source, returns "Stop" when there is no more left
    """
    global counter
    global reviews
    if counter<9:
      counter+=1
      return reviews[counter][0]
    return "Stop"

def mark_positive():
    """Marks review as positive"""
    global counter
    global reviews
    reviews[counter][1]=True

def mark_negative():
    """Marks review as negative"""
    global counter
    global reviews
    reviews[counter][1]=False
```
